# Investigación de una aplicación IHC actual: Duolingo

Ejercicio 1.5 de la Unidad 1. Se analiza la app Duolingo desde tres preguntas.


## Aplicación elegida

Duolingo es una app para aprender idiomas con lecciones cortas, rachas, puntos y ligas.


## Paradigma de interacción

Manipulación directa táctil, con apoyo multimodal (texto, audio, voz) y proactivo (notificaciones).


## Componentes de la IHC

Entrada (toque, micrófono, audífonos), proceso (motor de ejercicios), salida (pantalla, sonido, vibración), retroalimentación inmediata, modelo de usuario (nivel y racha) y contexto (hora e idioma).


## Objetivo que cumple mejor

La satisfacción: rachas, logros y ligas hacen que usar la app se sienta como un juego. También cumple usabilidad.


## Conclusión

Una buena interfaz no solo funciona, sino que sostiene la motivación del usuario.
